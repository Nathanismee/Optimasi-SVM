```python
# ===== SVM-Focused pipeline (sklearn 1.6 compatible) with visualizations =====
# Paste this cell and run. Uses existing X,y if present; otherwise loads 'fitness_dataset.csv'.
# Visualizations produced for each experiment's best estimator and for the final best_overall model:
# - ROC Curve (if probability or decision_function available)
# - Precision-Recall Curve
# - Confusion Matrix (counts + normalized)
# - Classification Report (printed)
# - Permutation feature importance (top features) when feasible
#
# NOTE: Some estimators (e.g., SGDClassifier) don't implement predict_proba. We attempt decision_function
# as fallback for ROC/PR; if neither available we skip ROC/PR for that estimator.

import time
import os
import random
import warnings
warnings.filterwarnings("ignore")

RANDOM_STATE = 42
os.environ['PYTHONHASHSEED'] = str(RANDOM_STATE)
random.seed(RANDOM_STATE)

import numpy as np
import pandas as pd
from copy import deepcopy
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, RandomizedSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.svm import SVC
from sklearn.linear_model import SGDClassifier, LogisticRegression
from sklearn.kernel_approximation import Nystroem
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier
from sklearn.metrics import (
    accuracy_score, classification_report, f1_score,
    roc_curve, auc, precision_recall_curve, average_precision_score,
    confusion_matrix
)
from sklearn.inspection import permutation_importance
import joblib

# Try to detect imblearn (SMOTE) availability
try:
    from imblearn.pipeline import Pipeline as ImbPipeline
    from imblearn.over_sampling import SMOTE
    IMB_AVAILABLE = True
    print("imblearn detected: SMOTE will be available for pipelines.")
except Exception:
    from sklearn.pipeline import Pipeline as ImbPipeline  # fallback to sklearn Pipeline
    SMOTE = None
    IMB_AVAILABLE = False
    print("imblearn not available: SMOTE disabled; will use class_weight='balanced' fallback.")

# ---------- Load or reuse X,y ----------
if 'X' in globals() and 'y' in globals():
    print("Using existing X,y in notebook.")
else:
    # adjust filename if different
    df = pd.read_csv('fitness_dataset.csv')
    # detect target column similarly to your notebook
    if 'target' in df.columns:
        target_col = 'target'
    elif 'label' in df.columns:
        target_col = 'label'
    else:
        target_col = df.columns[-1]
    X = df.drop(columns=[target_col]).copy()
    y = df[target_col].copy()
    print("Loaded X,y from fitness_dataset.csv with target:", target_col)

# ---------- Quick feature engineering (safe adds) ----------
def add_features_safe(Xdf):
    Xd = Xdf.copy()
    # BMI variants
    if set(['weight_kg','height_cm']).issubset(Xd.columns):
        if 'BMI' not in Xd.columns:
            Xd['height_m'] = Xd['height_cm'] / 100.0
            Xd['BMI'] = Xd['weight_kg'] / (Xd['height_m']**2 + 1e-9)
    elif set(['weight','height']).issubset(Xd.columns):
        if 'BMI' not in Xd.columns:
            Xd['height_m'] = Xd['height'] if Xd['height'].max() <= 3 else Xd['height']/100.0
            Xd['BMI'] = Xd['weight'] / (Xd['height_m']**2 + 1e-9)
    # derived
    if 'BMI' in Xd.columns and 'BMI_sq' not in Xd.columns:
        Xd['BMI_sq'] = Xd['BMI'] ** 2
    if set(['activity_index','nutrition_quality']).issubset(Xd.columns) and 'act_nut' not in Xd.columns:
        Xd['act_nut'] = Xd['activity_index'] * Xd['nutrition_quality']
    if set(['sleep_hours','activity_index']).issubset(Xd.columns) and 'sleep_act' not in Xd.columns:
        Xd['sleep_act'] = Xd['sleep_hours'] * Xd['activity_index']
    if set(['weight_kg','height_cm']).issubset(Xd.columns) and 'wt_ht_ratio' not in Xd.columns:
        Xd['wt_ht_ratio'] = Xd['weight_kg'] / (Xd['height_cm'] + 1e-6)
    return Xd

X = add_features_safe(X)

# ---------- Preprocessor (OneHotEncoder compatibility) ----------
num_cols = X.select_dtypes(include=['number']).columns.tolist()
cat_cols = X.select_dtypes(include=['object','category','bool']).columns.tolist()
print("Numeric cols:", len(num_cols), "| Categorical cols:", len(cat_cols))

# OneHotEncoder param compatibility
ohe_kwargs = {}
try:
    OneHotEncoder(sparse_output=False)
    ohe_kwargs['sparse_output'] = False
except TypeError:
    ohe_kwargs['sparse'] = False

numeric_transformer = Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())])
categorical_transformer = Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('ohe', OneHotEncoder(handle_unknown='ignore', **ohe_kwargs))])
preprocessor = ColumnTransformer([('num', numeric_transformer, num_cols), ('cat', categorical_transformer, cat_cols)], remainder='drop')

# ---------- Train/test split ----------
# protect against no-stratify if too few classes
try:
    strat = y
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, stratify=strat, random_state=RANDOM_STATE)
except Exception:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=RANDOM_STATE)
print("Train/test sizes:", X_train.shape, X_test.shape)

# ===== Visualisasi distribusi kelas SEBELUM SMOTE =====
print("\nDistribusi kelas SEBELUM SMOTE (y_train):")
print(y_train.value_counts())

plt.figure(figsize=(12,5))

# Bar chart
plt.subplot(1,2,1)
sns.countplot(x=y_train)
plt.title("Distribusi Kelas Sebelum SMOTE (Bar)")

# Pie chart
plt.subplot(1,2,2)
y_train.value_counts().plot.pie(autopct='%1.1f%%')
plt.title("Distribusi Kelas Sebelum SMOTE (Pie)")
plt.ylabel("")

plt.tight_layout()
plt.show()

# ---------- Utility: evaluate, save best & visualizer ----------
best_overall = {'name': None, 'acc': 0.0, 'model': None}

def try_get_score_values(estimator, X, method='auto'):
    """
    Return (y_score, score_type) where y_score is numeric values used for ROC/PR.
    Tries predict_proba -> decision_function -> predict (as fallback).
    score_type indicates 'proba', 'decision', or 'pred' (pred will not be useful for ROC/PR).
    """
    y_score = None
    score_type = None
    # If pipeline / ImbPipeline - call methods on pipeline directly (sklearn pipeline proxies)
    try:
        if hasattr(estimator, "predict_proba"):
            y_score = estimator.predict_proba(X)
            # If multiclass, take probability for positive label 1 if exists, else argmax? We return full if binary shaped
            if y_score.ndim == 2 and y_score.shape[1] >= 2:
                y_score = y_score[:, 1]
            else:
                # if only one column, keep it
                y_score = y_score.ravel()
            score_type = 'proba'
            return y_score, score_type
    except Exception:
        pass
    try:
        if hasattr(estimator, "decision_function"):
            y_score = estimator.decision_function(X)
            # if multiclass, try to reduce to one-dim by taking max or second dimension; best-effort
            if isinstance(y_score, np.ndarray) and y_score.ndim > 1:
                # for one-vs-rest, take column for label '1' if possible
                if y_score.shape[1] >= 2:
                    y_score = y_score[:, 1]
                else:
                    y_score = y_score.max(axis=1)
            score_type = 'decision'
            return y_score, score_type
    except Exception:
        pass
    try:
        # fallback to predictions (not good for curves, but useful to produce confusion)
        y_pred = estimator.predict(X)
        y_score = y_pred
        score_type = 'pred'
        return y_score, score_type
    except Exception:
        return None, None

def plot_evaluation(name, estimator, X_test, y_test, save_prefix=None, top_perm=15):
    """
    Produce ROC, PR, Confusion Matrix, classification report, and permutation importance plots.
    """
    if save_prefix is None:
        save_prefix = f"eval_{name}"
    # Predictions & basic report
    try:
        y_pred = estimator.predict(X_test)
    except Exception as e:
        print("Could not predict for", name, ":", e)
        return
    acc = accuracy_score(y_test, y_pred)
    print(f"\n== {name} | Test accuracy: {acc:.4f} ==")
    print(classification_report(y_test, y_pred, zero_division=0))

    # Confusion matrix (counts + normalized)
    cm = confusion_matrix(y_test, y_pred)
    labels = np.unique(np.concatenate([y_test.astype(str), y_pred.astype(str)]))
    fig, axes = plt.subplots(1,2, figsize=(12,5))
    sns.heatmap(cm, annot=True, fmt='d', ax=axes[0], cmap='Blues', xticklabels=labels, yticklabels=labels)
    axes[0].set_title(f'{name} Confusion Matrix (counts)')
    axes[0].set_xlabel('Predicted'); axes[0].set_ylabel('Actual')
    cm_norm = cm.astype('float') / (cm.sum(axis=1)[:, np.newaxis] + 1e-12)
    sns.heatmap(cm_norm, annot=True, fmt='.2f', ax=axes[1], cmap='Blues', xticklabels=labels, yticklabels=labels)
    axes[1].set_title(f'{name} Confusion Matrix (normalized)')
    axes[1].set_xlabel('Predicted'); axes[1].set_ylabel('Actual')
    plt.tight_layout()
    png = f"{save_prefix}_confusion.png"
    plt.savefig(png, dpi=150)
    plt.show()

    # ROC & Precision-Recall (attempt)
    y_score, stype = try_get_score_values(estimator, X_test, method='auto')
    if y_score is None:
        print(f"Skipping ROC/PR for {name} (no score available).")
    else:
        # If y_score is discrete predictions (dtype int) treat accordingly
        if stype in ('proba','decision'):
            # ROC
            try:
                fpr, tpr, _ = roc_curve(y_test, y_score)
                roc_auc = auc(fpr, tpr)
                plt.figure(figsize=(7,6))
                plt.plot(fpr, tpr, lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
                plt.plot([0,1],[0,1], linestyle='--', lw=1, color='gray')
                plt.xlim([0.0,1.0]); plt.ylim([0.0,1.05])
                plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
                plt.title(f'ROC Curve - {name}')
                plt.legend(loc='lower right')
                plt.grid(alpha=0.3)
                plt.tight_layout()
                png = f"{save_prefix}_roc.png"
                plt.savefig(png, dpi=150)
                plt.show()
            except Exception as e:
                print("ROC plotting failed for", name, ":", e)

            # PR curve
            try:
                precision, recall, _ = precision_recall_curve(y_test, y_score)
                ap = average_precision_score(y_test, y_score)
                plt.figure(figsize=(7,6))
                plt.plot(recall, precision, lw=2, label=f'AP = {ap:.3f}')
                plt.xlabel('Recall'); plt.ylabel('Precision')
                plt.title(f'Precision-Recall Curve - {name}')
                plt.legend(loc='lower left')
                plt.xlim([0.0,1.0]); plt.ylim([0.0,1.05])
                plt.grid(alpha=0.3)
                plt.tight_layout()
                png = f"{save_prefix}_pr.png"
                plt.savefig(png, dpi=150)
                plt.show()
            except Exception as e:
                print("PR plotting failed for", name, ":", e)
        else:
            # stype == 'pred' -- not useful for ROC; skip curves
            print(f"Score available for {name} is discrete predictions; ROC/PR skipped.")

    # Permutation importance (works with any predict; but may be slow)
    try:
        # permutation importance expects the estimator to accept raw X (pipeline does)
        print("Computing permutation importance (this can take a while)...")
        r = permutation_importance(estimator, X_test, y_test, n_repeats=16, random_state=RANDOM_STATE, n_jobs=-1)
        imp_mean = r.importances_mean
        # build feature names if pipeline supports get_feature_names_out on preprocessor
        feat_names = None
        try:
            # if estimator is pipeline with 'preproc' step that's a ColumnTransformer
            if hasattr(estimator, 'named_steps') and 'preproc' in estimator.named_steps:
                # try to get feature names from preprocessor
                pre = estimator.named_steps['preproc']
                try:
                    feat_names = pre.get_feature_names_out()
                    feat_names = [str(fn) for fn in feat_names]
                except Exception:
                    # try numeric + onehot get_feature_names_out
                    num_names = num_cols
                    if len(cat_cols) > 0:
                        cat_tr = pre.named_transformers_.get('cat')
                        if cat_tr is not None and hasattr(cat_tr.named_steps.get('ohe'), 'get_feature_names_out'):
                            ohe = cat_tr.named_steps.get('ohe')
                            try:
                                cat_names = list(ohe.get_feature_names_out(cat_cols))
                            except Exception:
                                cat_names = [f"{c}_ohe_*" for c in cat_cols]
                        else:
                            cat_names = [f"{c}_ohe_*" for c in cat_cols]
                    else:
                        cat_names = []
                    feat_names = list(num_names) + list(cat_names)
            else:
                # not pipeline with preproc; attempt estimator.feature_names_in_
                if hasattr(estimator, 'feature_names_in_'):
                    feat_names = list(estimator.feature_names_in_)
        except Exception:
            feat_names = None

        if feat_names is None:
            # fallback: generic indices
            feat_names = [f"f{i}" for i in range(len(imp_mean))]

        df_imp = pd.DataFrame({'feature': feat_names[:len(imp_mean)], 'importance_mean': imp_mean}).sort_values('importance_mean', ascending=False)
        display(df_imp.head(top_perm))
        # plot top k
        topk = df_imp.head(top_perm).iloc[::-1]
        plt.figure(figsize=(8, max(4, 0.25*len(topk))))
        plt.barh(topk['feature'], topk['importance_mean'])
        plt.xlabel('Permutation importance (mean)')
        plt.title(f'Top {top_perm} features - {name}')
        plt.tight_layout()
        png = f"{save_prefix}_permimp.png"
        plt.savefig(png, dpi=150)
        plt.show()
    except Exception as e:
        print("Permutation importance failed for", name, ":", e)

    # return metrics
    return acc

def evaluate_and_save(name, fitted_pipeline, plot=True):
    # print evaluation, save model if best, optionally plot
    try:
        y_pred = fitted_pipeline.predict(X_test)
    except Exception as e:
        print(f"Prediction failed for {name}: {e}")
        return 0.0
    acc = accuracy_score(y_test, y_pred)
    print(f"\n== {name} | Test accuracy: {acc:.4f} ==")
    print(classification_report(y_test, y_pred, zero_division=0))
    global best_overall
    if acc > best_overall['acc']:
        best_overall = {'name': name, 'acc': acc, 'model': deepcopy(fitted_pipeline)}
        try:
            joblib.dump(fitted_pipeline, f"best_svm_{name}.joblib")
            print("Saved model to", f"best_svm_{name}.joblib")
        except Exception as e:
            print("Could not save model:", e)
    if plot:
        try:
            plot_evaluation(name, fitted_pipeline, X_test, y_test, save_prefix=f"eval_{name}")
        except Exception as e:
            print("Plotting failed for", name, ":", e)
    return acc

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

# ---------- EXPERIMENT A: Wide SVC tuning (SMOTE if available; else class_weight balanced) ----------
print("\n>>> Experiment A: Wide SVC RandomizedSearchCV")
svc = SVC(probability=True, random_state=RANDOM_STATE)

param_dist = {
    'model__C': [1e-3,1e-2,1e-1,0.5,1,2,5,10,50,100,300],
    'model__gamma': ['scale','auto', 1e-3,1e-2,1e-1,0.5,1],
    'model__kernel': ['rbf','poly','sigmoid'],
    'model__degree': [2,3,4]
}

PipelineClass = ImbPipeline if IMB_AVAILABLE else Pipeline
steps = [('preproc', preprocessor), ('kbest', SelectKBest(score_func=f_classif, k='all'))]
if IMB_AVAILABLE:
    steps.append(('smote', SMOTE(random_state=RANDOM_STATE)))
else:
    svc = SVC(probability=True, class_weight='balanced', random_state=RANDOM_STATE)
steps.append(('model', svc))
pipe_svc = PipelineClass(steps)

rs_svc = RandomizedSearchCV(pipe_svc, param_distributions=param_dist, n_iter=45, scoring='accuracy', cv=cv, n_jobs=-1, random_state=RANDOM_STATE, verbose=1)
t0 = time.time()
rs_svc.fit(X_train, y_train)
t1 = time.time()
print("SVC tuning finished in", int(t1-t0), "s. Best CV:", rs_svc.best_score_)
best_pipe_svc = rs_svc.best_estimator_
evaluate_and_save('svc_wide', best_pipe_svc, plot=True)

# ---------- EXPERIMENT B: RF feature importance -> SVC on top-k features ----------
print("\n>>> Experiment B: RF feature importance -> SVC on top-k")
rf_feat = RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1)
pipe_rf_feat = Pipeline([('preproc', deepcopy(preprocessor)), ('rf', rf_feat)])
pipe_rf_feat.fit(X_train, y_train)
# get transformed feature names if possible
feat_names = None
try:
    # numeric feature names
    num_names = num_cols
    # cat feature names from fitted OHE
    if len(cat_cols) > 0:
        cat_transformer = pipe_rf_feat.named_steps['preproc'].named_transformers_.get('cat')
        ohe_step = cat_transformer.named_steps.get('ohe')
        cat_names = list(ohe_step.get_feature_names_out(cat_cols))
    else:
        cat_names = []
    feat_names = num_names + cat_names
except Exception:
    feat_names = None

importances = pipe_rf_feat.named_steps['rf'].feature_importances_
k = min(30, len(importances))
top_idx = np.argsort(importances)[-k:]
print("Selecting top k:", k, "features for SVM.")

sel = SelectKBest(score_func=f_classif, k=k)
svc2 = SVC(probability=True, random_state=RANDOM_STATE)
steps2 = [('preproc', deepcopy(preprocessor)), ('kbest', sel)]
if IMB_AVAILABLE:
    steps2.append(('smote', SMOTE(random_state=RANDOM_STATE)))
else:
    svc2 = SVC(probability=True, class_weight='balanced', random_state=RANDOM_STATE)
steps2.append(('model', svc2))
pipe_sel_svc = ImbPipeline(steps2)

param_dist2 = {
    'model__C': [0.1,1,5,10,50,100],
    'model__gamma': ['scale','auto', 1e-2, 1e-1, 0.5],
    'model__kernel': ['rbf','poly'],
    'model__degree': [2,3]
}
rs_sel = RandomizedSearchCV(pipe_sel_svc, param_distributions=param_dist2, n_iter=25, scoring='accuracy', cv=cv, n_jobs=-1, random_state=RANDOM_STATE, verbose=1)
rs_sel.fit(X_train, y_train)
print("Best params (sel->svc):", rs_sel.best_params_)
best_sel_pipe = rs_sel.best_estimator_
evaluate_and_save('svc_selk', best_sel_pipe, plot=True)

# ---------- EXPERIMENT C: Kernel approx (Nystroem) + SGDClassifier ----------
print("\n>>> Experiment C: Nystroem kernel approx + SGDClassifier")
n_components = 300  # reduce if memory constrained
pipe_kapprox = Pipeline([
    ('preproc', deepcopy(preprocessor)),
    ('kbest', SelectKBest(score_func=f_classif, k='all')),
    ('featmap', Nystroem(kernel='rbf', gamma=0.1, n_components=n_components, random_state=RANDOM_STATE)),
    ('clf', SGDClassifier(loss='hinge', max_iter=5000, tol=1e-4, random_state=RANDOM_STATE))
])
param_dist_k = {'clf__alpha': [1e-4,5e-4,1e-3,5e-3], 'featmap__gamma':[0.01,0.05,0.1,0.2]}
rs_kapprox = RandomizedSearchCV(pipe_kapprox, param_distributions=param_dist_k, n_iter=12, scoring='accuracy', cv=cv, n_jobs=-1, random_state=RANDOM_STATE, verbose=1)
rs_kapprox.fit(X_train, y_train)
best_kapprox = rs_kapprox.best_estimator_
print("Best params (kapprox):", rs_kapprox.best_params_)
# Note: SGDClassifier doesn't have predict_proba; ROC/PR will attempt decision_function fallback.
evaluate_and_save('kapprox_sgd', best_kapprox, plot=True)

# ---------- EXPERIMENT D: Bagging SVC (estimator=...) ----------
print("\n>>> Experiment D: Bagging SVC to reduce variance")
# use the best_sel_pipe as base estimator's 'model' if available; else default SVC
try:
    # best_sel_pipe may be an imblearn pipeline; get underlying model object
    base_svc = deepcopy(best_sel_pipe.named_steps['model'])
except Exception:
    base_svc = SVC(probability=True, C=10, gamma='scale', kernel='rbf', random_state=RANDOM_STATE)

# BaggingClassifier uses 'estimator' arg in sklearn >=1.4
bag = BaggingClassifier(estimator=base_svc, n_estimators=10, max_samples=0.8, max_features=0.9, n_jobs=-1, random_state=RANDOM_STATE)
steps_b = [('preproc', deepcopy(preprocessor)), ('kbest', SelectKBest(score_func=f_classif, k='all'))]
if IMB_AVAILABLE:
    steps_b.append(('smote', SMOTE(random_state=RANDOM_STATE)))
steps_b.append(('bag', bag))
pipe_bag = ImbPipeline(steps_b)
pipe_bag.fit(X_train, y_train)
evaluate_and_save('bagging_svc', pipe_bag, plot=True)

# ---------- Final summary ----------
print("\n=== Best overall SVM result ===")
print(best_overall)
if best_overall['model'] is not None:
    print("You can load it with: joblib.load('best_svm_{}.joblib')".format(best_overall['name']))
    # produce final visualization (again) for the saved best model (non-deepcopy)
    try:
        # if saved, reload the file to ensure consistent artifact
        fname = f"best_svm_{best_overall['name']}.joblib"
        if os.path.exists(fname):
            final_model = joblib.load(fname)
            print("\nProducing final plots for saved best model:", fname)
            plot_evaluation(f"best_{best_overall['name']}", final_model, X_test, y_test, save_prefix=f"eval_best_{best_overall['name']}")
        else:
            plot_evaluation(f"best_{best_overall['name']}", best_overall['model'], X_test, y_test, save_prefix=f"eval_best_{best_overall['name']}")
    except Exception as e:
        print("Final plotting failed:", e)

        # ===== Visualisasi distribusi kelas SETELAH SMOTE (FIXED & AMAN) =====
if IMB_AVAILABLE:
    from sklearn.impute import SimpleImputer

    smote_vis = SMOTE(random_state=RANDOM_STATE)

    # 1. Encoding kategorikal
    X_temp = pd.get_dummies(X_train, drop_first=True)

    # 2. HANDLE MISSING VALUE (WAJIB sebelum SMOTE)
    imputer = SimpleImputer(strategy='median')
    X_temp_imputed = imputer.fit_transform(X_temp)

    # 3. SMOTE
    X_res, y_res = smote_vis.fit_resample(X_temp_imputed, y_train)

    print("\nDistribusi kelas SETELAH SMOTE:")
    print(pd.Series(y_res).value_counts())

    plt.figure(figsize=(12,5))

    # Bar chart
    plt.subplot(1,2,1)
    sns.countplot(x=y_res)
    plt.title("Distribusi Kelas Setelah SMOTE (Bar)")

    # Pie chart
    plt.subplot(1,2,2)
    pd.Series(y_res).value_counts().plot.pie(autopct='%1.1f%%')
    plt.title("Distribusi Kelas Setelah SMOTE (Pie)")
    plt.ylabel("")

    plt.tight_layout()
    plt.show()

    # ===== BONUS: Perbandingan sebelum vs sesudah =====
    before_counts = y_train.value_counts()
    after_counts = pd.Series(y_res).value_counts()

    df_compare = pd.DataFrame({
        'Before SMOTE': before_counts,
        'After SMOTE': after_counts
    }).fillna(0)

    df_compare.plot(kind='bar', figsize=(8,5))
    plt.title("Perbandingan Distribusi Kelas Sebelum vs Setelah SMOTE")
    plt.ylabel("Jumlah Data")
    plt.xticks(rotation=0)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

else:
    print("SMOTE tidak tersedia, visualisasi setelah SMOTE tidak dapat ditampilkan.")

# End of cell
```
